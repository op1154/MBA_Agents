```python
import os
from google.colab import userdata

from langchain_openai import ChatOpenAI

from langchain_core.output_parsers import StrOutputParser

from langchain_core.prompts import ChatPromptTemplate

from langchain_core.chat_history import InMemoryChatMessageHistory

from langchain_core.runnables.history import RunnableWithMessageHistory



# load_dotenv() # Removed as we are using Colab secrets

api_key = userdata.get("OPENAI_API_KEY") # Changed to use Colab secrets



modelo = ChatOpenAI(

    model="gpt-5-nano",

    temperature=0.3,

    api_key=api_key

)
prompt_sugestao = ChatPromptTemplate.from_messages([

    ("system", "Você é um assistente de vendas especializado em produtos para ambientes domésticos."),

    ("placeholder", "{historico}"),

    ("human", "{query}")

])


cadeia = prompt_sugestao | modelo | StrOutputParser()



memoria = {}

sessao = "Langchain_sessao"


def historico_de_mensagens(sessao : str):

    if sessao not in memoria:

        memoria[sessao] = InMemoryChatMessageHistory()

    return memoria[sessao]


lista_perguntas = [

    "Qual produto é mais indicado para o ambiente de cozinha?",

    "Qual produto pode ser indicado como complementar para o ambiente?"

]

cadeia_mem = RunnableWithMessageHistory(

    runnable=cadeia,

    get_session_history=historico_de_mensagens,

    input_messages_key="query",

    history_messages_key="historico",

)

for pergunta in lista_perguntas:

    resposta = cadeia_mem.invoke(

        input={"query": pergunta},

        config={"configurable": {"session_id": sessao}}

    )
    print(f"Pergunta Usuario: {pergunta}\n Resposta: {resposta}\n")
```
