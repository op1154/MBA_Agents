```python
from pathlib import Path

import os

from dotenv import load_dotenv
from openai import OpenAI
from google.colab import userdata

#Coloquei estes somente no .ENV do VSCode
#dotenv_path = Path(__file__).resolve().parent / ".env​
#load_dotenv(dotenv_path=dotenv_path, override=True)

# Carrega a API key dos secrets do Colab
# 1. Clique no ícone '🔑 Secrets' na barra lateral esquerda.
# 2. Adicione um novo segredo com o nome 'OPENAI_API_KEY' e seu valor.
# 3. Certifique-se de que 'Notebook access' esteja ativado para este notebook.
os.environ['OPENAI_API_KEY'] = userdata.get('OPENAI_API_KEY')

admin_api_key = os.getenv("OPENAI_API_KEY")

# Adicionando a verificação novamente para garantir que a chave foi carregada
if not admin_api_key:
    raise RuntimeError("OPENAI_API_KEY não foi encontrada. Por favor, configure-a nos Secrets do Colab.")

if not admin_api_key.startswith("sk-"):
    raise RuntimeError("OPENAI_API_KEY inválida. Verifique o valor nos Secrets do Colab.")

qual_produto = "TV de 55 polegadas"

ambiente = "Sala de estar de 6 metros quadrados"

modelo = "u8600f"

prompt = f"""Crie um escritivo de suporte para o produto {qual_produto}, para o uso em {ambiente}​

            detalhando o modelo {modelo} e descreva as melhores características."""


cliente = OpenAI(api_key=admin_api_key)

response = cliente.chat.completions.create(

    model="gpt-5-nano",

    messages=[

        {"role": "system", "content": "Você é um assistente de suporte especializado em produtos eletrônicos."},

        {"role": "user", "content": prompt},

    ],

)
print(response.choices[0].message.content)
```
